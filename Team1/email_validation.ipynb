```python
# Email Validation Project - Excel Upload Version (Colab)

import re
import pandas as pd
from google.colab import files

# ─────────────────────────────────────────
# STEP 1: Upload your Excel file
# ─────────────────────────────────────────

print("📂 Please upload your Excel file...")
uploaded = files.upload()   # A file picker button will appear

filename = list(uploaded.keys())[0]
df = pd.read_excel(filename, header=0)

# Check what pandas is reading
print("\n🔍 Preview of your file:")
print(df.head())
print(f"\n✅ File loaded! Rows found: {len(df)}")
print(f"   Columns: {list(df.columns)}\n")

# ─────────────────────────────────────────
# STEP 2: Column name (matches your header)
# ─────────────────────────────────────────

EMAIL_COLUMN = "Email Id"   # ← matches your Excel header exactly

if EMAIL_COLUMN not in df.columns:
    print(f"❌ Column '{EMAIL_COLUMN}' not found.")
    print(f"   Available columns: {list(df.columns)}")
    print("   👆 Copy one of these and update EMAIL_COLUMN above")
    raise ValueError("Column not found — check the name above")

# ─────────────────────────────────────────
# STEP 3: Validation Logic
# ─────────────────────────────────────────

ALLOWED_PROVIDERS = {"gmail", "hotmail", "yahoo", "google", "microsoft"}
def validate_email(email):
    # Handle missing / empty values
    if pd.isna(email) or str(email).strip() == "":
        return "❌ Empty / Missing"

    email = str(email).strip()

    # Check capital letter at start
    if email[0].isupper():
        return "❌ Must not start with a capital letter"

    pattern = r'^[a-z][a-zA-Z0-9._%+-]*@([a-zA-Z]+)\.com$'
    match = re.match(pattern, email)

    if not match:
        if "@" not in email:
            return "❌ Missing @"
        if not email.split("@")[0][0].isalpha():
            return "❌ Must start with a letter"
        if not email.endswith(".com"):
            return "❌ Must end with .com"
        return "❌ Invalid Format"

    provider = match.group(1).lower()

    if provider not in ALLOWED_PROVIDERS:
        return f"❌ Invalid Provider ({provider})"

    return "✅ Valid"
# ─────────────────────────────────────────
# STEP 4: Apply to all emails
# ─────────────────────────────────────────

df["Validation Result"] = df[EMAIL_COLUMN].apply(validate_email)

# ─────────────────────────────────────────
# STEP 5: Show Results
# ─────────────────────────────────────────

print("========== EMAIL VALIDATION RESULTS ==========\n")
print(df[["Student Name", EMAIL_COLUMN, "Validation Result"]].to_string(index=False))

print("\n---------- SUMMARY ----------")
print(df["Validation Result"].value_counts().to_string())

valid_count   = (df["Validation Result"] == "✅ Valid").sum()
invalid_count = len(df) - valid_count
print(f"\nTotal: {len(df)}  |  Valid: {valid_count}  |  Invalid: {invalid_count}")

# ─────────────────────────────────────────
# STEP 6: Download results as Excel
# ─────────────────────────────────────────

output_file = "email_validation_results.xlsx"
df.to_excel(output_file, index=False)
files.download(output_file)

print(f"\n✅ Results downloaded as '{output_file}'")
```
