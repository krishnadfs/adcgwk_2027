1. What is meant by Classification?

Classification is a machine learning technique used to divide data into different
categories or classes.

It is a supervised learning method where the model learns from existing data and
predicts the class for new data.

In classification, input data is given to the model and based on previous patterns,
it predicts the output category.

Example in Student Data Analysis:

Input:
- Attendance
- CGPA
- Technical Skills
- Marks

Output:
- Good Performance
- Average Performance
- Poor Performance

Classification helps to predict student performance and supports better decision
making.

2. What are the methods in Classification?


There are different classification methods used in machine learning:

1. Logistic Regression:
It is used for classification problems where output is usually two classes.
Example: Pass or Fail.

2. Decision Tree:
It uses a tree-like structure to make decisions based on conditions.
Example:
If CGPA is high and attendance is good, then performance is good.

3. Random Forest:
It combines multiple decision trees and gives better prediction accuracy.

4. Support Vector Machine (SVM):
It separates different classes by finding the best boundary between them.

5. K-Nearest Neighbor (KNN):
It predicts the class based on similar data points.

6. Naive Bayes:
It uses probability to classify the data into different categories.

*   List item



3. How does Classification work? (Mathematical Formulation)

Classification works by learning the relationship between input data and output
categories.

  The student data is represented as input features:

X = (Attendance, CGPA, Technical Skills)

The output is represented as:

Y = Student Performance Category


The model learns the relationship:

Y = f(X)

  where:

X = Input data
Y = Predicted output


In Logistic Regression, the mathematical formula is:

Z = b0 + b1x1 + b2x2 + .... + bnxn

  Where:

b0 = bias
b1,b2...bn = weights
x1,x2...xn = input values
After calculating Z, probability is found using sigmoid function:

P = 1 / (1 + e^-Z)
Based on the probability value, the model assigns the student to a class.

Example:

Attendance = 85%
CGPA = 8.2

Prediction:

Student Performance = Good

```python
# ===============================
# TASK 6 : CGPA PERFORMANCE ANALYSIS
# Google Colab
# ===============================

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Upload Excel file
uploaded = files.upload()

# Get uploaded file name automatically
file_name = list(uploaded.keys())[0]

# Read Excel file
excel = pd.ExcelFile(file_name)

print("Available Sheets:")
print(excel.sheet_names)

# Read the allStudents sheet
df = pd.read_excel(file_name, sheet_name="allStudents")

# Remove missing values
df = df.dropna(subset=["CGPA"])

# Convert CGPA to numeric
df["CGPA"] = pd.to_numeric(df["CGPA"], errors="coerce")

df = df.dropna(subset=["CGPA"])

print("\nTotal Students:", len(df))

# -----------------------------
# Data Characteristics
# -----------------------------
print("\n===== CGPA Statistics =====")

print("Mean :", round(df["CGPA"].mean(),2))
print("Variance :", round(df["CGPA"].var(),2))
print("Median :", round(df["CGPA"].median(),2))
print("Minimum :", df["CGPA"].min())
print("Maximum :", df["CGPA"].max())

print("\nPercentiles")
print(df["CGPA"].quantile([0.25,0.50,0.75,0.90]))

# -----------------------------
# Performance Groups
# -----------------------------
def performance(cgpa):
    if cgpa >= 9:
        return "Excellent"
    elif cgpa >= 8:
        return "Very Good"
    elif cgpa >= 7:
        return "Good"
    elif cgpa >= 6:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["CGPA"].apply(performance)

print("\nPerformance Count")
print(df["Performance"].value_counts())

# -----------------------------
# Pivot Table
# -----------------------------
pivot = pd.pivot_table(
    df,
    values="CGPA",
    index="Performance",
    aggfunc=["count","mean","min","max"]
)

print("\n===== Pivot Table =====")
print(pivot)

# -----------------------------
# Histogram
# -----------------------------
plt.figure(figsize=(7,5))
plt.hist(df["CGPA"], bins=10)
plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Number of Students")
plt.show()

# -----------------------------
# Bar Chart
# -----------------------------
plt.figure(figsize=(7,5))
df["Performance"].value_counts().plot(kind="bar")
plt.title("Student Performance Groups")
plt.xlabel("Performance")
plt.ylabel("Number of Students")
plt.show()

# -----------------------------
# Box Plot
# -----------------------------
plt.figure(figsize=(5,6))
plt.boxplot(df["CGPA"])
plt.title("CGPA Box Plot")
plt.ylabel("CGPA")
plt.show()

print("\nTask 6 Completed Successfully")
```

```python
# ============================================
# TASK-1: Data Cleaning & Data Validation
# ============================================

import pandas as pd
from google.colab import files

# Upload Excel File
uploaded = files.upload()

# Read uploaded file
file_name = list(uploaded.keys())[0]
df = pd.read_excel(file_name)

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)

# ==================================================
# Basic Information
# ==================================================

print("\nDataset Info:")
print(df.info())

print("\nFirst 5 Rows:")
print(df.head())

# ==================================================
# Remove Duplicate Records
# ==================================================

duplicate_count = df.duplicated().sum()
print("\nDuplicate Records:", duplicate_count)

df = df.drop_duplicates()

# ==================================================
# Remove Extra Spaces
# ==================================================

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.strip()

# ==================================================
# Standardize Text Columns
# ==================================================

text_columns = [
    "Board Name",
    "School Name",
    "School Type",
    "School Location",
    "College Name",
    "College Type",
    "College Location",
    "Group"
]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].str.title()

# ==================================================
# Missing Values
# ==================================================

print("\nMissing Values")
print(df.isnull().sum())

# Fill missing numeric values with median
numeric_cols = df.select_dtypes(include=['number']).columns

for col in numeric_cols:
    df[col].fillna(df[col].median(), inplace=True)

# Fill missing text values with Unknown
text_cols = df.select_dtypes(include='object').columns

for col in text_cols:
    df[col].fillna("Unknown", inplace=True)

# ==================================================
# Email Validation
# ==================================================

email_pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

invalid_email = df[
    ~df["Email"].astype(str).str.match(email_pattern)
]

print("\nInvalid Email Count:", len(invalid_email))

# ==================================================
# Convert Numeric Columns
# ==================================================

numeric_columns = [
    "10th Marks",
    "Intermediate (+2) Marks",
    "Current CGPA"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# ==================================================
# Marks Validation
# ==================================================

invalid_10th = df[
    (df["10th Marks"] < 0) |
    (df["10th Marks"] > 100)
]

print("\nInvalid 10th Marks:", len(invalid_10th))

invalid_inter = df[
    (df["Intermediate (+2) Marks"] < 0) |
    (df["Intermediate (+2) Marks"] > 100)
]

print("Invalid Intermediate Marks:", len(invalid_inter))

# ==================================================
# CGPA Validation
# ==================================================

invalid_cgpa = df[
    (df["Current CGPA"] < 0) |
    (df["Current CGPA"] > 10)
]

print("Invalid CGPA:", len(invalid_cgpa))

# ==================================================
# Board Validation
# ==================================================

valid_boards = ["Cbse", "State"]

invalid_board = df[
    ~df["Board Name"].isin(valid_boards)
]

print("Invalid Board Names:", len(invalid_board))

# ==================================================
# School Type Validation
# ==================================================

valid_school = ["Private", "Govt"]

invalid_school = df[
    ~df["School Type"].isin(valid_school)
]

print("Invalid School Types:", len(invalid_school))

# ==================================================
# College Type Validation
# ==================================================

valid_college = ["Private", "Govt"]

invalid_college = df[
    ~df["College Type"].isin(valid_college)
]

print("Invalid College Types:", len(invalid_college))

# ==================================================
# Summary
# ==================================================

print("\n==============================")
print("DATA VALIDATION SUMMARY")
print("==============================")

print("Total Records :", len(df))
print("Duplicate Records Removed :", duplicate_count)
print("Invalid Emails :", len(invalid_email))
print("Invalid 10th Marks :", len(invalid_10th))
print("Invalid Intermediate Marks :", len(invalid_inter))
print("Invalid CGPA :", len(invalid_cgpa))
print("Invalid Board Names :", len(invalid_board))
print("Invalid School Types :", len(invalid_school))
print("Invalid College Types :", len(invalid_college))

# ==================================================
# Save Cleaned Dataset
# ==================================================

output_file = "Cleaned_Student_Data.xlsx"

df.to_excel(output_file, index=False)

print("\nCleaned file saved as:", output_file)

files.download(output_file)
```

This Task-1 code performs data cleaning and data validation on the uploaded student dataset to ensure the data is accurate, consistent, and ready for analysis. It starts by reading the Excel file uploaded in Google Colab, displaying the dataset information, and removing duplicate records. The code then trims unnecessary spaces from text columns, standardizes categorical values such as board name, school type, and college type, and checks for missing values. Missing numeric values are replaced with the median of their respective columns, while missing text values are filled with "Unknown" to maintain data completeness.

After cleaning the data, the code validates important fields such as email addresses, 10th marks, Intermediate (+2) marks, and Current CGPA by checking whether they fall within the expected ranges and formats. It also verifies that board names and institution types contain only valid values (CBSE/STATE and Private/Govt). Finally, the program displays a validation summary showing any invalid records found, saves the cleaned dataset as a new Excel file, and automatically downloads it. This cleaned dataset serves as the foundation for building pivot tables and performing further analysis in the subsequent tasks.

```python
# ==========================================================
# TASK-5
# Combined Influence of 10th & Intermediate on Current CGPA
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

# ----------------------------------------------------------
# Upload Excel File
# ----------------------------------------------------------

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name)

print("Dataset Loaded Successfully")
print("Shape :", df.shape)

print("\nColumns")
print(df.columns.tolist())

# ----------------------------------------------------------
# Data Cleaning
# ----------------------------------------------------------

df.columns = df.columns.str.strip()

df["10th Marks"] = pd.to_numeric(df["10th Marks"], errors="coerce")
df["Intermediate Marks"] = pd.to_numeric(df["Intermediate Marks"], errors="coerce")
df["CGPA"] = pd.to_numeric(df["CGPA"], errors="coerce")

df.dropna(subset=["10th Marks","Intermediate Marks","CGPA"], inplace=True)

print("\nRows after cleaning :", len(df))

# ----------------------------------------------------------
# Data Validation
# ----------------------------------------------------------

print("\nMissing Values\n")
print(df.isnull().sum())

print("\nDuplicate Records :", df.duplicated().sum())

# ----------------------------------------------------------
# Previous Academic Score
# 50% weight from 10th
# 50% weight from Intermediate
# ----------------------------------------------------------

df["10th Percentage"] = (df["10th Marks"]/600)*100
df["Intermediate Percentage"] = (df["Intermediate Marks"]/1000)*100

df["Previous Academic Score"] = (
    df["10th Percentage"]*0.5
    +
    df["Intermediate Percentage"]*0.5
)

print("\nPrevious Academic Score Created Successfully")

# ----------------------------------------------------------
# Correlation
# ----------------------------------------------------------

print("\nCorrelation Matrix\n")

print(df[
    [
        "10th Percentage",
        "Intermediate Percentage",
        "Previous Academic Score",
        "CGPA"
    ]
].corr())

# ----------------------------------------------------------
# Pivot Table 1
# Average CGPA by Board
# ----------------------------------------------------------

pivot1 = pd.pivot_table(
    df,
    values="CGPA",
    index="Board Name",
    aggfunc=np.mean
)

print("\nAverage CGPA by Board")
print(pivot1)

# ----------------------------------------------------------
# Pivot Table 2
# School Type
# ----------------------------------------------------------

pivot2 = pd.pivot_table(
    df,
    values="CGPA",
    index="School Type",
    aggfunc=np.mean
)

print("\nAverage CGPA by School Type")
print(pivot2)

# ----------------------------------------------------------
# Pivot Table 3
# College Type
# ----------------------------------------------------------

pivot3 = pd.pivot_table(
    df,
    values="CGPA",
    index="College Type",
    aggfunc=np.mean
)

print("\nAverage CGPA by College Type")
print(pivot3)

# ----------------------------------------------------------
# Pivot Table 4
# Group Wise
# ----------------------------------------------------------

pivot4 = pd.pivot_table(
    df,
    values="CGPA",
    index="Group",
    aggfunc=np.mean
)

print("\nAverage CGPA by Group")
print(pivot4)

# ----------------------------------------------------------
# Statistics
# ----------------------------------------------------------

print("\nDescriptive Statistics\n")

print(df[
    [
        "10th Percentage",
        "Intermediate Percentage",
        "Previous Academic Score",
        "CGPA"
    ]
].describe())

# ----------------------------------------------------------
# Scatter Plot
# ----------------------------------------------------------

plt.figure(figsize=(8,6))

plt.scatter(
    df["Previous Academic Score"],
    df["CGPA"]
)

plt.title("Previous Academic Score vs CGPA")

plt.xlabel("Previous Academic Score")

plt.ylabel("Current CGPA")

plt.grid(True)

plt.show()

# ----------------------------------------------------------
# Histogram
# ----------------------------------------------------------

plt.figure(figsize=(8,5))

plt.hist(df["CGPA"], bins=10)

plt.title("Distribution of CGPA")

plt.xlabel("CGPA")

plt.ylabel("Number of Students")

plt.show()

# ----------------------------------------------------------
# Top Students
# ----------------------------------------------------------

top_students = df.sort_values(
    by="CGPA",
    ascending=False
)[
    [
        "Name",
        "CGPA",
        "10th Marks",
        "Intermediate Marks",
        "Previous Academic Score"
    ]
].head(10)

print("\nTop 10 Students")
print(top_students)

# ----------------------------------------------------------
# Final Correlation
# ----------------------------------------------------------

corr = df["Previous Academic Score"].corr(df["CGPA"])

print("\n===================================================")
print("TASK-5 RESULT")
print("===================================================")

print("Correlation :", round(corr,3))

if corr >= 0.70:
    print("Strong positive influence of previous academic performance on CGPA.")

elif corr >= 0.40:
    print("Moderate positive influence of previous academic performance on CGPA.")

elif corr >= 0.20:
    print("Weak positive influence of previous academic performance on CGPA.")

else:
    print("Very little or no influence observed.")

print("\nTask-5 Completed Successfully.")
```

```python
import pandas as pd
import io

# 1. Load the data
# (You can export your Google Sheet as a CSV or paste the raw text data here)
raw_data = """Name,Corrected Name (Camel Case),Email Id ,GROUP,Roll Number,ChatGpt Account,VS Code Status,Git Account Status,CGPA,Personal Laptop Acces
ALEKHYA MOTURU,Alekhya Moturu,alekhyamoturu630@gmail.com,BCA,2467640203,Active,YTC,YTC,8.07,Yes
ANKITA GONAPA,Ankita Gonapa,ankithagonapa@gmail.com,BCA,2467640173,Active ,YTC,YTC ,8.41,Yes
BHUMIKA PRABHAKAR SONAWANE,Bhumika Prabhakar Sonawane,bhumikabps19@gmail.com,BCA,2467640124,Active ,Active ,Active ,8.24,Yes
BINITA NATH,Binita Nath,binitanath319@gmail.com,BCA,2467640219,Active,Active ,Active,8.4,Yes
HASINI PINNAMARAJU,Hasini Pinnamaraju,pinnamarajuhasini@gmail.com,BSC-B,2467640184,Active,Active,Active,9.21,yes
SWATHI LAGUDU,Swathi Lagudu,laguduswathi360@gmail.com,BSC-C,2467640286,active ,YTC,active ,9.5,No
SAI BHANU TEJITA MANEPALLI,Sai Bhanu Tejita Manepalli,tejita0906@gmail.com,BCA-DS,2467640253,Active,Active,Active ,8.82,Yes
"""

# Read data into a DataFrame
df = pd.read_csv(io.StringIO(raw_data.strip()))

# 2. Data Cleaning
# Clean column names by removing trailing spaces
df.columns = df.columns.str.strip()

# Standardize text columns to uppercase/lowercase to avoid duplicate categories
df['GROUP'] = df['GROUP'].str.strip().str.upper()
df['VS Code Status'] = df['VS Code Status'].str.strip().str.upper()
df['Git Account Status'] = df['Git Account Status'].str.strip().str.upper()

# Convert CGPA to a numeric type, forcing missing or invalid strings to NaN
df['CGPA'] = pd.to_numeric(df['CGPA'], errors='coerce')

# Drop students with missing CGPA values for accurate analysis
df = df.dropna(subset=['CGPA'])

print("--- ANALYSIS RESULTS ---")

# 3. Analyze Feature 1: Academic Group vs. Performance
print("\n### Average CGPA by Academic Group ###")
group_performance = df.groupby('GROUP')['CGPA'].mean().reset_index()
print(group_performance.to_string(index=False))

# 4. Analyze Feature 2: Technical Tool Status vs. Performance
print("\n### Average CGPA by VS Code Activation Status ###")
vscode_performance = df.groupby('VS Code Status')['CGPA'].mean().reset_index()
print(vscode_performance.to_string(index=False))

print("\n### Average CGPA by Git Activation Status ###")
git_performance = df.groupby('Git Account Status')['CGPA'].mean().reset_index()
print(git_performance.to_string(index=False))

# 5. Create a combined metric: Full Environment Setup vs. Incomplete Setup
def check_readiness(row):
    if row['VS Code Status'] == 'ACTIVE' and row['Git Account Status'] == 'ACTIVE':
        return 'Fully Active Setup'
    else:
        return 'Incomplete (YTC) Setup'

df['Technical_Readiness'] = df.apply(check_readiness, axis=1)

print("\n### How Tech Onboarding Influences Overall Performance ###")
readiness_analysis = df.groupby('Technical_Readiness')['CGPA'].mean().reset_index()
print(readiness_analysis.to_string(index=False))
```
